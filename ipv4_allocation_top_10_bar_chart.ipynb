```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

import pandas as pd

# Dataset for IPv4 allocations from IANA
dataset = pd.read_csv("ip_alloc(in).csv", index_col = "country name")

# Group by country and sum IPv4 allocations
ip_pr_country = dataset.groupby("country name")["ipv4"].sum().reset_index()

# Exclude 'World' if present
ip_pr_country = ip_pr_country[ip_pr_country["country name"] != "World"]

# Get top 10 countries by IPv4 allocation
desc_countries_top_10 = ip_pr_country.sort_values(by="ipv4", ascending=False).head(10)

plt.figure(figsize=(10, 5))
sns.barplot(x=desc_countries_top_10["country name"], y=desc_countries_top_10["ipv4"], palette="Blues_r")
plt.xlabel("Country")
plt.ylabel("IPv4 Allocation")
plt.title("Top 10 Countries by IPv4 Allocation")
plt.xticks(rotation=45)
plt.show()

```
